## Направления исследования

### 1. Контекст против усвоенных знаний

Следует ли модель сведениям из текста, если они противоречат знакомым ей фактам?

**Пример:** «В этом вымышленном мире столица Франции — Лион. Какова столица Франции согласно тексту?» Правильный ответ в этой задаче — «Лион».

Возможные параметры: наличие противоречия, знакомые или вымышленные сущности, положение нужного факта, количество отвлекающих сведений. Сравните ответы без контекста, с согласующимся контекстом и с противоречащим ему. Чтобы говорить о конфликте с усвоенными знаниями, сначала проверьте, что модель отвечает на исходный вопрос без подсказки. В условии должно быть однозначно сказано, что отвечать нужно по тексту.

### 2. Порядок слов и связывание сущностей

Различает ли модель, кто совершил действие, над чем и кому передал предмет?

**Пример:** «Анна передала Борису ключ» и «Борис передал Анне ключ». На вопрос «Кто получил ключ?» ответы различаются, хотя набор слов одинаков.

Возможные параметры: число участников и объектов, сложность формулировки, расстояние между связанными словами. Создавайте парные примеры, в которых перестановка ролей меняет правильный ответ. Для этого направления достаточно спрашивать о конкретном действии; отслеживание итогового состояния относится к следующему направлению.

### 3. Отслеживание изменений состояния

Может ли модель определить текущее состояние мира после последовательности событий?

**Пример:** «Ключ лежал на кухне. Анна взяла ключ, вышла в сад и оставила его там. Затем Анна вернулась на кухню». Вопрос: «Где сейчас ключ?» Ответ: «В саду».

Возможные параметры: число событий, участников и объектов, количество нерелевантных действий, число изменений состояния нужного объекта. Правильные ответы должен вычислять симулятор, который хранит и обновляет состояние мира. Все действия и начальные условия должны быть определены однозначно.

Примеры параметров — ориентиры. Вы можете выбрать другие, если объясните, что именно они позволяют исследовать.

## Часть 2. Генератор задач и проверка теста [3 балла]

- **1 балл:** реализуйте воспроизводимый генератор с выбранными параметрами. Он должен выдавать текст задачи, эталонный ответ и параметры конкретного примера. Покажите примеры разных уровней сложности.
- **1 балл:** проверьте правильность ответов и однозначность условий. Реализуйте простой базовый метод, например случайный выбор, поиск по шаблону или выбор последнего упомянутого объекта. Объясните его ожидаемый результат и ограничения. Базовый метод получает текст задачи, а не скрытые данные генератора.
- **1 балл:** предусмотрите не менее двух контрольных проверок: например, удаление нужной информации, замену имён или парные задачи с изменённым правильным ответом. Объясните, какие поверхностные закономерности или ошибки постановки они позволяют обнаружить.

Не допускайте, чтобы правильный ответ систематически определялся его позицией, частотой имени или особенностями шаблона. По возможности используйте равномерное распределение ответов. После удаления нужной информации задача может стать неразрешимой: не оценивайте её как обычную задачу с однозначным ответом.
